## 1. What are capital markets?

**Capital markets** are where governments, companies and investors **raise and trade money** through financial products.

- A **company** that needs money can sell **shares** (stocks) or **borrow** by issuing bonds.
- A **government** borrows money by issuing **government bonds**.
- **Investors** (banks, funds, people) buy these products hoping to earn a return.
- Money also moves between countries, so currencies are traded too: **foreign exchange (FX)**.

**Where Murex fits:** banks use Murex's **MX.3** platform to trade these products, measure their risk, and process the deals.

### My questions and answers

**Q: When we say "governments," do we mean cities?**
Not usually. In capital markets, *government* mostly means a **country's national government** (for example the US, France or Lebanon). Cities can borrow too, through *municipal bonds*, but they are a smaller part of the market.

**Q: Who are investors?**
Anyone with money to invest: banks, pension funds, insurance companies, companies, governments and ordinary people. They **invest** their money so that it grows over time.

**Q: Do investors give their money for free?**
No. They expect their money to come back with more. There are **two main ways** this happens:

| | **Lending** (bonds) | **Owning** (stocks) |
|---|---|---|
| What the investor does | **Lends** money | **Buys a piece** of the company |
| What the investor gets | **Interest** (payment for using the money) + the money back at the end | A share of the **profits** (dividends) + gains if the **price goes up** |
| Is the return promised? | ✅ Yes, agreed in advance | ❌ No, nothing is guaranteed |
| Risk | Lower | Higher |

- **Bond investors are lenders.** The company or government **owes** them their interest and their money back.
- **Stock investors are owners.** They share the company's good years *and* its bad years.

**Key idea: the risk–return trade-off.** A higher possible return comes with a higher risk.

**Quick check:** *If I own shares in a company and it has a bad year, does it still owe me money?*
**No.** As an owner, I share the losses. The company only owes money to its **lenders** (bondholders), not to its owners.

## 2. The three asset classes in this project

| Asset class | What it is | The key number |
|---|---|---|
| **Equities** (stocks) | A share of ownership in a company | **Return** in % |
| **Rates** (government bonds) | A loan to a government | **Yield** in %, with changes in **basis points** |
| **FX** (foreign exchange) | The price of one currency in another | **Exchange rate**, with changes in % |

An **index** (like the S&P 500) tracks a group of stocks together. It shows how "the market" is doing overall.

### My questions and answers

**My summary of the three asset classes:**
- **Equities (stocks):** the investor **buys part of a company**. The reward is **not promised**: if the company does well, the investor gains; if it does badly, the investor can lose.
- **Rates (government bonds):** the investor **lends** money to a government. The one who gives money is **lending**; the one who takes it (the government) is **borrowing**. The government pays interest and returns the money at the end.
- **FX (foreign exchange):** the **price of one currency in another**, for example how many US dollars one euro costs.

**Q: What does "an index shows how the market is doing" mean?**
Think of a **class average**. Instead of checking 30 students' grades one by one, a teacher looks at one number: the average.
An **index** does the same for stocks:
- The **S&P 500** follows **500 of the biggest US companies** (Apple, Microsoft, Amazon…).
- Instead of watching 500 prices, investors watch **one number**.
- If the S&P 500 **rises 1%**, those companies rose about **1% on average**.

Difference from a class average: **bigger companies count more**, so Apple moves the index more than a small company.

## 3. Equities: what is a return?

A **return** is how much the price changed, as a percentage.

**daily return (%) = (today's price ÷ yesterday's price − 1) × 100**

**Example:** an index goes from 5,000 to 5,050 → (5050 ÷ 5000 − 1) × 100 = **+1%**

```python
# Example: 5 days of prices for a stock index
prices = [5000, 5050, 5025, 5100, 5000]

# Each day's return compares it with the day before,
# so the first day has no return.
for i in range(1, len(prices)):
    daily_return = (prices[i] / prices[i - 1] - 1) * 100
    print(f"Day {i+1}: price {prices[i]}  ->  return {daily_return:+.2f}%")
```

## 4. Bonds, yields and basis points

A **bond** is a loan that can be bought and sold. When I buy a bond, I am **lending** money (not owning, like with stocks).

| Term | Meaning | Example |
|---|---|---|
| **Issuer** | Who borrows the money | French government |
| **Face value (par)** | The amount paid back at the end | €1,000 |
| **Coupon** | Fixed yearly interest, as a % of face value | 5% → €50 per year |
| **Maturity** | When the loan ends | 5 years |

So with this bond I receive €50 every year, and my €1,000 back after 5 years.

**Yield** is the return I actually earn if I buy the bond at today's market price and keep it until maturity. The coupon never changes, but the price does, so the yield changes.

A **basis point (bp)** is 0.01%. So 100 bp = 1%.
Example: a yield going from 5.00% to 5.25% has risen by **25 bp**.

### My questions and answers

**Q: If the coupon is fixed, how can my return change?**
Because I don't always pay €1,000 for the bond. If I buy it cheaper, I still get the same €50 and the same €1,000 at the end, so my return is higher. If I pay more, my return is lower. The coupon is fixed; the yield depends on the price I pay.

**Q: Why do traders say "basis points" instead of "%"?**
Because "%" can be confusing. If a yield is 5% and someone says "it rose by 1%", it could mean 6% (one percentage point) or 5.05% (1% of 5). "It rose by 100 bp" can only mean 6%. No confusion.

```python
# Annual coupon of a bond
face_value = 1000
coupon_rate = 0.05

annual_coupon = face_value * coupon_rate
print(f"Annual coupon: €{annual_coupon:.2f}")

# A yield moves from 4.80% to 5.15%: how many basis points?
old_yield = 4.80
new_yield = 5.15

change_bp = (new_yield - old_yield) * 100
print("Raw result:", change_bp)
print(f"Change: {round(change_bp)} bp")
```

**Q: Why 35.00000000000005?**
Computers store decimal numbers in binary, and numbers like 4.80 and 5.15 can't be stored exactly. So tiny errors appear. This is called **floating-point error**. That's why I use `round()`. In finance this matters: when testing pricing code, we compare numbers with a small **tolerance** instead of expecting an exact match.

### Why bond prices and yields move in opposite directions

Imagine I own a bond paying a 5% coupon. Then interest rates in the market go up, and **new bonds now pay 6%**.
Nobody wants my 5% bond at full price anymore, because they can get 6% elsewhere. So to sell it, I must **lower the price** until the buyer's return is about 6%.

- Rates go **up** → old bonds become less attractive → their **price goes down**
- Rates go **down** → old bonds become more attractive → their **price goes up**

**Price and yield always move in opposite directions.**

Simple example: a bond that pays €1,050 in one year (€1,000 face value + €50 coupon).
Its price today = 1050 ÷ (1 + yield)

```python
# A bond that pays €1,050 in one year (face value + coupon)
cash_at_maturity = face_value + annual_coupon

for y in [0.04, 0.05, 0.06]:
    price = cash_at_maturity / (1 + y)
    print(f"Yield {y * 100:.0f}%  ->  price €{price:.2f}")
```

**Q: What does this output show me?**
At a 5% yield, the price is exactly €1,000 (yield = coupon). When the yield goes up to 6%, the price drops to about €990.57. When the yield goes down to 4%, the price rises to about €1,009.62. Same bond, same payments, only the price changes.

**Q: Who loses when rates go up?**
Anyone already holding bonds, because the market value of their bonds falls. This is called **interest rate risk**, and banks measure it every day (Murex does this too).

## 5. The yield curve

The **yield curve** shows the yields of government bonds with different maturities (1 year, 2 years, 5 years, 10 years, 30 years) on one chart.

- **Normal curve (upward sloping):** longer loans pay more. If I lend for 30 years, I want more reward than for 1 year, because more can go wrong in 30 years.
- **Inverted curve (downward sloping):** short-term yields are *higher* than long-term yields. This is unusual. It often means the market expects interest rates to fall in the future, often because of a slowdown.

The numbers below are examples, not real market data.

```python
import matplotlib.pyplot as plt

maturities = [1, 2, 5, 10, 30]              # years
normal_curve = [3.0, 3.3, 3.7, 4.1, 4.4]    # yields in %
inverted_curve = [5.0, 4.7, 4.3, 4.0, 3.9]  # yields in %

plt.figure(figsize=(8, 5))
plt.plot(maturities, normal_curve, marker="o", label="Normal (upward sloping)")
plt.plot(maturities, inverted_curve, marker="o", label="Inverted (downward sloping)")
plt.xlabel("Maturity (years)")
plt.ylabel("Yield (%)")
plt.title("Yield curve: normal vs inverted (example numbers)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# The "10y - 2y spread" in basis points: a common way to describe the curve's shape
spread_normal = round((normal_curve[3] - normal_curve[1]) * 100)
spread_inverted = round((inverted_curve[3] - inverted_curve[1]) * 100)
print(f"10y - 2y spread, normal curve:   {spread_normal:+} bp")
print(f"10y - 2y spread, inverted curve: {spread_inverted:+} bp")
```

### My questions and answers

**Q: Why would anyone accept less for lending for longer (inverted curve)?**
Because investors expect rates to go down. If I think rates will fall, I want to lock in today's long-term yield before it drops, even if it's a bit lower than the short-term one.

**Q: Is an inverted curve a sign of recession?**
In the US, an inverted curve has come before most recessions in the past decades, so people watch it closely. But it's a warning sign, not a guarantee.

**Q: What is the "10y − 2y spread"?**
The 10-year yield minus the 2-year yield, usually in basis points. Positive = normal curve. Negative = inverted. One number that describes the shape of the whole curve.

## 6. FX: how a currency quote works

A quote like **EUR/USD = 1.10** means: **1 euro costs 1.10 US dollars.**

- The first currency (EUR) is the **base currency**: always 1 unit.
- The second currency (USD) is the **quote currency**: how much of it you pay for 1 unit of the base.

If EUR/USD goes from 1.10 to **1.15**, one euro now buys more dollars:
- the **euro got stronger**
- the **dollar got weaker**

FX moves are often counted in **pips**. For EUR/USD, 1 pip = 0.0001 (the 4th decimal).

```python
eur_usd_before = 1.10
eur_usd_after = 1.15
amount_eur = 1000

print(f"€{amount_eur} = ${amount_eur * eur_usd_before:.2f} at EUR/USD {eur_usd_before}")
print(f"€{amount_eur} = ${amount_eur * eur_usd_after:.2f} at EUR/USD {eur_usd_after}")

# Change in %
change_pct = (eur_usd_after / eur_usd_before - 1) * 100
print(f"EUR/USD change: {change_pct:+.2f}%")

# The same rate seen from the other side: USD/EUR = 1 / (EUR/USD)
print(f"USD/EUR: {1 / eur_usd_before:.4f} -> {1 / eur_usd_after:.4f}")

# Change in pips
pips = round((eur_usd_after - eur_usd_before) * 10000)
print(f"Move: {pips} pips")
```
